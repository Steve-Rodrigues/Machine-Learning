I will be using the diabetes data set from sklearn to perform regularization and compare the learning curves between the linear regression using polynomially boosted features against the ridge regression model which was regularized to reduce the variance and add a small bias

```python
from sklearn.linear_model import LinearRegression, RidgeCV #using cv to find the optimal alpha value to use as a penalty
from sklearn.datasets import load_diabetes
from sklearn.model_selection import learning_curve
import matplotlib.pyplot as plt
import numpy as np
```

```python
X,y = load_diabetes(return_X_y=True)
```

```python
plt.plot(X,y, 'b.')
plt.show()
```

```python
training_sizes, train_scores, val_scores = learning_curve(LinearRegression(),X,y,train_sizes=np.linspace(0.1,1.0,50),cv=5,scoring='neg_mean_squared_error')
```

```python
training_sizes_means = -train_scores.mean(axis=1)
val_sizes_means = -val_scores.mean(axis=1) #means for each size-- since k=5 we have 5 scores for each size becasue each fold needs to run
plt.plot(training_sizes, training_sizes_means, 'b-')
plt.plot(training_sizes, val_sizes_means, 'r-')
plt.xlabel('Training Size')
plt.ylabel('Mean Squared Error')
plt.show()

```
