I will be performing binary classifcation using the built in iris flower data set to determine whether a flower is setosa species or not

```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LogisticRegression
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
import matplotlib.pyplot as plt
```

```python
data = load_iris(as_frame=True)
df = pd.DataFrame(data.data)
y = np.astype(np.array(data.target == 0),int) #only going to be using setosa for the binary classifier
X = df.copy() 
```

```python
X_train, X_test, y_train, y_test = train_test_split(X,y, stratify=y,train_size=0.8,random_state=42)
```

```python
log_model = LogisticRegression()
log_model.fit(X_train, y_train)
```

```python
log_model.predict_proba(X_train) #shows the probability of each instance being the setosa species vs not (y=1 vs y=0)
```

```python
cm = confusion_matrix(y_true=y_train, y_pred=log_model.predict(X_train))
ConfusionMatrixDisplay(cm).plot()
plt.show()
```

```python
testing_pred = log_model.predict(X_test)
cm = confusion_matrix(y_true=y_test, y_pred=testing_pred)
ConfusionMatrixDisplay(cm).plot()
plt.show()
```
