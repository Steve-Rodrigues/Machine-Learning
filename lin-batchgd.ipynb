I will be writing the algorithm for linear regression using the batch gradient descent technique. The batch gradient descent technique is going to go through all of the instances in the feature matrix every time we get a gradient, so it is quite expensive as more and more rows are added.

```python
import numpy as np 
from sklearn.preprocessing import add_dummy_feature
```

Below is the batch gradient descent function to find the values for the parameter column vector given that best fit to the feature matrix also given. The end goal is to find the values for all the parameters that are going to give the closest predictions to the true labels and then be used to generalize to other testing data

```python
def fit_batch_gradient_descent(X, y, parameters):
    X = add_dummy_feature(X) #adds the dummy 1 column to go into all of the parameters during matrixmult(X0*O0)
    gradient_vector = np.zeros_like(parameters, dtype=float) #holds all of the gradients for each parameter
    epoches = 100000 #number of iterations we allow through the data set
    learning_rate = 0.1 #the step we take in the direction of the gradient to minimize cost
    tolerance = 1e-8 #threshold for rate of cost decrease
    m=len(X)
    curr_cost = np.mean(((X@parameters)-y)**2)
    for epoch in range(epoches):
        #current parameter we are minimizing is j
        for j in range(len(gradient_vector)):
            gradient_sum = 0 #holds the sum of the error scaled by the left over feature from derivative
            for i in range(len(X)):
                curr_feature = X[i][j] #gives the feature that is currently scaling the error
                error = (X[i]@parameters)-y[i] #error of prediction from current instance
                gradient_sum += (curr_feature*error)
            gradient_vector[j] = 2*gradient_sum / m
        #now lets update the parameters given the gradients in the vector
        parameters = parameters - (learning_rate*gradient_vector)
        #check if the cost decrease enough
        new_cost = np.mean(((X@parameters)-y)**2)
        if curr_cost < new_cost:
            print(f"The gradient increased instead of decreasing after {epoch} epoches. Check the learning rate")
        if curr_cost-new_cost <= tolerance:
            print(f"Converged after {epoch} epoches.")
            break
        curr_cost = new_cost
    return parameters,X #returns the parameters that best minimize the cost function 
```

Testing this out on the diabetes data set from sklearn

```python
from sklearn.datasets import load_diabetes
from sklearn.preprocessing import StandardScaler

X,y = load_diabetes(return_X_y=True)
y = np.reshape(y,shape=(442,1))
scaled_X = StandardScaler().fit_transform(X) #returns the scaled features 
parameters = np.zeros(shape=(11,1)) #starts as all 0 so each error will be pretty high to start
fitted_params,X = fit_batch_gradient_descent(scaled_X,y,parameters)
mse = np.mean((X@fitted_params-y)**2)

```

Checking the values I got for the mse from batch gradient descent with the values I get for the parameter using the normal equation because there are only 10 features

```python
X,y = load_diabetes(return_X_y=True)
X = add_dummy_feature(X)
parameters = np.linalg.inv(X.T@X)@X.T@y
mse = np.mean((X@parameters-y)**2)
mse #we got the same as batch gradient descent
```

Using batch gradient descent and the normal equation I ended up getting the same parameters and got the same MSE of 2859 on the training data
